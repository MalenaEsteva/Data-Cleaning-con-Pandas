# Práctica de Data Cleaning


## Objetivo

El objetivo de este ejercicio es aplicar técnicas de **limpieza de datos** en Pandas.

Este ejercicio cubre la manipulación de strings (utilizando métodos de strings), el uso de expresiones regulares, la importancia de la limpieza y la conversión a `datetime`

## Contexto del Conjunto de Datos

Hemos cargado un DataFrame que contiene información de Presidentes de EE. UU. La columna `'Born'` (Nacimiento) contiene la fecha y, entre paréntesis, la ciudad o ubicación, como por ejemplo: `February 22, 1732[b] (Westmoreland County, Virginia)`.

## Instrucciones

- De la columna `'President'` Crea una columna para el nombre 'firstname' y otra para el apellido (lastname) de los presidentes eliminando la referencias a las notas al pie que encuentres. Puedes utilizar un patron de Regular Expression para realizar esta tarea
- La columna `'Born'` contiene referencias entre corchetes (ej. `[b]`) y texto adicional entre paréntesis (ej. `(Westmoreland County, Virginia)`). Debemos eliminar todo este texto para quedarnos solo con la fecha limpia.
- Convierte la columna `'Born'` al Tipo `datetime`
- Extrae de la columna  `'Born'` el año de nacimiento y genera una nueva columna denominada `'Year Born'`


### Carga del conjunto de datos

Se importa la librería Pandas y se carga el archivo `presidents.csv` en un DataFrame. Luego se muestran las primeras filas para observar el formato de las columnas.


```python
import pandas as pd

df = pd.read_csv("../data/presidents.csv")

df.head()
```

### Revisión de los nombres de las columnas

Se muestran los nombres de las columnas del DataFrame para confirmar cómo están escritos antes de trabajar con ellos.

```python
df.columns
```

### Limpieza de la columna President

Se eliminan de la columna `'President'` las posibles referencias a notas al pie, como `[a]`, mediante una expresión regular. El patrón busca un corchete de apertura, una o más letras o números y un corchete de cierre, y lo reemplaza por un texto vacío.

```python
df['President'] = df['President'].str.replace(r'\[\w+\]', '', regex=True)

df['President'].head(10)
```

### Creación de la columna firstname

Se crea la columna `'firstname'` a partir de `'President'`. Mediante una expresión regular se elimina todo el texto que aparece desde el primer espacio en adelante, de modo que solo queda la primera palabra, que corresponde al nombre.

```python
df['firstname'] = df['President'].str.replace(r'\s.*', '', regex=True)

df[['President', 'firstname']].head(10)
```

### Creación de la columna lastname

Se crea la columna `'lastname'` a partir de `'President'`. Mediante una expresión regular se elimina todo el texto que aparece hasta el último espacio, de modo que solo queda la última palabra, que corresponde al apellido.

```python
df['lastname'] = df['President'].str.replace(r'.*\s', '', regex=True)

df[['President', 'firstname', 'lastname']].head(10)
```

### Limpieza de la columna Born

Se eliminan de la columna `'Born'` las referencias a notas al pie entre corchetes, como `[a]`, y el posible texto adicional entre paréntesis, como `(Westmoreland County, Virginia)`. De esta forma, la columna queda solo con la fecha.

```python
df['Born'] = df['Born'].str.replace(r'\[\w+\]', '', regex=True)
df['Born'] = df['Born'].str.replace(r'\s*\(.*\)', '', regex=True)

df['Born'].head(10)
```

### Conversión de la columna Born a datetime

Se convierte la columna `'Born'` al tipo `datetime` mediante la función `pd.to_datetime()`, para que Pandas reconozca los valores como fechas y no como texto. Luego se verifica el tipo de dato de la columna.

```python
df['Born'] = pd.to_datetime(df['Born'])

print(df['Born'].dtype)
df['Born'].head(10)
```

### Extracción del año de nacimiento

Se crea la columna `'Year Born'` extrayendo el año de la columna `'Born'` mediante el accesor `.dt.year`, disponible para columnas de tipo `datetime`. Finalmente, se muestran las columnas trabajadas para verificar el resultado.

```python
df['Year Born'] = df['Born'].dt.year

df[['President', 'firstname', 'lastname', 'Born', 'Year Born']].head(10)
```
